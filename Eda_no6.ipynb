### 1.1 Introduction and Problem Statement
Carbon emissions from individual’s daily behaviors is a growing contributor of 
global climate change, but personal level behavioral data is not being used 
properly in sustainability analysis.  
Malaysia is especially relevant in this context because, according to the World Bank, Malaysia produces more CO2 emissions than almost half of OECD countries and 3.0 MT/capita more than the world average of 5.0 MT/capita (Malaysia Population Research Hub, 2022).  
One main factor is the country’s dependence on private vehicles due to underdeveloped public transportation infrastructure, this will increase transport related carbon outputs.  
Lastly, the methodologies for carbon footprint calculations are still evolving and it is emerging as an important tool for greenhouse gas management (Pandey et al., 2011).  

### 1.2 Main Objective
This study analyses daily behavioral records from 200 individuals from transport mode, electricity consumption, dietary habits, renewable energy usage, waste generation, and pro environmental actions.  
This study aims to identify which lifestyle factors most significantly influence personal carbon footprints and to evaluate whether eco friendly behaviours can meaningfully reduce high emission habits.

### 1.3 Key Analytical Questions
- Which behavioural variables are most strongly associated with daily carbon footprint? 
(Correlation analysis)
- Can an individual's daily carbon footprint be predicted from their behavioural inputs? 
(Linear regression)
- Do individuals naturally cluster into different carbon behaviour profiles based on their
 daily habits? (K Means clustering)
- What mix of behaviours most accurately classifies an individual's
 carbon impact level as Low, Medium, or High? (Decision Tree / Random Forest classification)

Before transforming any columns, we first check for outliers in the original numerical columns using the IQR (interquartile range) method, where any value beyond 1.5×IQR from the 25th/75th percentile is flagged.

```python
outlier_cols = ['distance_km', 'electricity_kwh', 'screen_time_hours',
                'waste_generated_kg', 'eco_actions']

for col in outlier_cols:
    q1, q3 = df[col].quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    n_outliers = ((df[col] < lower) | (df[col] > upper)).sum()
    print(col, ': ', n_outliers, ' outlier(s) outside [', round(lower,2), ',', round(upper,2), ']')
```

**Description:**  
Only a small number of outliers are flagged (a handful of records in `electricity_kwh` and `waste_generated_kg`, none in the other columns), and the maximum values are still physically realistic for a single day (e.g., 12.27 kWh electricity, 1.51 kg waste). These likely represent genuinely high-usage days rather than data entry errors, so we choose **not to remove** them. Deleting real high-usage days would bias the dataset toward underestimating carbon footprints and would specifically remove some of the records that later get classified as "High" carbon impact, exactly the records the classification task in Section 6 needs to learn from. We therefore keep all 1400 records and proceed to feature engineering below.

We going to convert electricity_kwh and renewable_usage_pct into non_renewable_kwh and renewable_kwh

```python
df["renewable_electricity_kwh"] = df["electricity_kwh"] * df["renewable_usage_pct"] / 100
df["nonrenewable_electricity_kwh"] = df["electricity_kwh"] - df["renewable_electricity_kwh"]
df.drop(columns=["electricity_kwh", "renewable_usage_pct"], inplace=True)
df
```

We going make enumerate the food_type

```python
df['food_type'] = df['food_type'].map({'Veg': 0, 'Mixed': 1, 'Non-Veg': 2})
df
```

We going make enumerate the transport_mode

```python
df['transport_mode'] = df['transport_mode'].map(
    {'Bike': 0, 'Walk': 1, 'EV': 2, 'Bus': 3, 'Car': 4})
df
```

### 6.0 K-Means Clustering (Behaviour Profiles)
This shows our third analytical question which is do individuals naturally cluster into different carbon behaviour profiles based on their daily habits?

Clustering measures distance between points, so it only works correctly on variables that are genuinely numeric on a real scale (like kWh or km). We leave out `transport_mode` and `food_type` here, even though they were encoded as numbers in Section 4, because those numbers are just labels for categories (e.g. Car=4, Bike=0), not real quantities, K-Means would wrongly treat Car and Bus as "close together" just because 4 and 3 are close, which has no real meaning. We also standardize the remaining features using StandardScaler before running k-means, since distance_km ranges up to 22 while eco_actions only ranges 0-5, following the same approach as in our clustering lab.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

cluster_cols = ['distance_km', 'screen_time_hours', 'waste_generated_kg',
                'eco_actions', 'renewable_electricity_kwh', 'nonrenewable_electricity_kwh']
Xc = df[cluster_cols]

ss = StandardScaler()
Xc_scaled = ss.fit_transform(Xc)
print(Xc.head())
print(Xc_scaled[:5])
```

**Description:**  
Xc shows the original feature values, while Xc_scaled shows the same rows after standardization, each column now has mean 0 and unit variance, so no single feature like distance_km, which has a naturally larger range, dominates the distance calculation that k-means relies on. Only the 6 genuinely continuous variables are included here.

We try a few values of k and look at the inertia (sum of squared distances of points to their assigned cluster center) to get a sense of how clustering quality changes as k increases.

```python
inertia_list = []
k_range = range(2, 7)
for k in k_range:
    model_k = KMeans(k, verbose=0, n_init='auto', random_state=42)
    model_k.fit(Xc_scaled)
    inertia_list.append(model_k.inertia_)

plt.plot(list(k_range), inertia_list, "bo-")
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Plot for K-Means')
plt.show()
```

**Description:**  
Inertia decreases steadily as k increases, without one single sharp "elbow". We choose **k=3** rather than picking purely off this plot, since our dataset already has a natural 3-way grouping concept (carbon_impact_level with Low, Medium, High), and k=3 lets us check whether the clusters discovered from behaviour alone resemble this real-world grouping.

```python
model = KMeans(3, verbose=0, n_init='auto', random_state=42)
model.fit(Xc_scaled)
```

```python
print(model.cluster_centers_.shape)
model.cluster_centers_
```

```python
print(model.labels_.shape)
model.labels_
```

```python
model.inertia_
```

(cluster_centers_) gives us the centre of each of the 3 clusters in standardized feature space, (labels_) assigns each of the 1400 records to one of the 3 clusters (0, 1, or 2), and (inertia_) is the total sum of squared distances of points to their assigned cluster centre, lower generally means tighter, more well-separated clusters.

```python
kmeans_labels = pd.DataFrame(model.labels_)
df['cluster'] = kmeans_labels
df[['carbon_footprint_kg', 'carbon_impact_level', 'cluster']].head(10)
```

```python
# Check how well the unsupervised clusters line up with the real carbon_impact_level labels
pd.crosstab(df['cluster'], df['carbon_impact_level'])
```

**Description:**  
We can observe that cluster 2 captures the majority of the high impact records, 155 out of 194, and this cluster also has the highest average nonrenewable_electricity_kwh, which is 6.17 kWh, and the highest average carbon_footprint_kg, 9.88 kg, among the three clusters. Cluster 2 also has the lowest average renewable_electricity_kwh (0.54 kWh), so this group is best described as "high non-renewable energy use". Cluster 0 has a higher average carbon footprint (7.51 kg) than cluster 1 (6.34 kg), and uses more non-renewable electricity (3.77 kWh vs 2.02 kWh) and more screen time (6.49 hours vs 5.21 hours), while cluster 1 uses more renewable electricity (4.33 kWh) and takes slightly more eco actions on average. This tells us k-means, using only genuine behavioural quantities with no access to the actual footprint number, still rediscovers a grouping that lines up closely with the real-world High/Medium/Low impact categories, especially for High impact behaviour, and the main factor separating the groups is non-renewable electricity use.

```python
X = df[['transport_mode', 'distance_km', 'food_type',
        'screen_time_hours', 'waste_generated_kg', 'eco_actions',
        'renewable_electricity_kwh', 'nonrenewable_electricity_kwh']]
y = df['carbon_footprint_kg']
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, random_state=42)
```

```python
# Compare regression models
from sklearn.linear_model import Ridge, Lasso
from sklearn.metrics import mean_absolute_error, r2_score
import pandas as pd

models = {
    'Linear Regression': LinearRegression(),
    'Ridge (alpha=1)': Ridge(alpha=1.0, random_state=42),
    'Ridge (alpha=10)': Ridge(alpha=10.0, random_state=42),
    'Lasso (alpha=0.1)': Lasso(alpha=0.1, random_state=42),
    'Decision Tree (depth=10)': DecisionTreeRegressor(max_depth=10, random_state=42),
}

results = []
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    results.append({
        'Model': name,
        'R²': r2_score(y_test, y_pred),
        'MAE': mean_absolute_error(y_test, y_pred),
    })

results_df = pd.DataFrame(results).round(4)
print(results_df)
```

```python
# Bar plot comparing regression models
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# R² plot (higher is better)
colors_r2 = ['#2ecc71' if v >= 0.95 else '#f39c12' if v >= 0.85 else '#e74c3c'
             for v in results_df['R²']]
axes[0].bar(results_df['Model'], results_df['R²'], color=colors_r2)
axes[0].set_ylabel('R² Score')
axes[0].set_title('R² Score by Model')
axes[0].tick_params(axis='x', rotation=15)

# MAE plot (lower is better)
colors_mae = ['#2ecc71' if v <= 0.4 else '#f39c12' if v <= 0.6 else '#e74c3c'
              for v in results_df['MAE']]
axes[1].bar(results_df['Model'], results_df['MAE'], color=colors_mae)
axes[1].set_ylabel('MAE (kg CO₂)')
axes[1].set_title('MAE by Model')
axes[1].tick_params(axis='x', rotation=15)

plt.tight_layout()
plt.show()
```
